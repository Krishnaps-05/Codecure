```python
# ============================================================
# INFORMATION RETRIEVAL (IR) - TF, IDF AND TF-IDF
# CELL 1: IMPORT REQUIRED LIBRARIES
# ============================================================

import re
import math
import pandas as pd
import numpy as np

from collections import Counter
from google.colab import files

# Stopwords
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# Stemming
from nltk.stem import PorterStemmer

print("All required libraries imported successfully!")
```

```python
# ============================================================
# CELL 2: UPLOAD THE 5 TEXT DOCUMENTS
# ============================================================

print("Select your 5 social-media text files.")

uploaded = files.upload()

print("\nUploaded files:")
for filename in uploaded.keys():
    print("✓", filename)

print(f"\nTotal files uploaded: {len(uploaded)}")
```

```python
# ============================================================
# CELL 3: READ AND CHECK THE 5 TEXT DOCUMENTS
# ============================================================

documents = {}

for filename in uploaded.keys():
    with open(filename, "r", encoding="utf-8") as file:
        documents[filename] = file.read()

print("=" * 70)
print("DOCUMENT COLLECTION")
print("=" * 70)

print(f"\nTotal documents: {len(documents)}")

for i, (filename, text) in enumerate(documents.items(), start=1):
    print(f"\nDocument {i}: {filename}")
    print(f"Number of characters: {len(text)}")
    print(f"Approximate number of words: {len(text.split())}")

    print("\nFirst 200 characters:")
    print(text[:200])
    print("-" * 70)
```

```python
# ============================================================
# CELL 4: PREPROCESSING
# ============================================================
# Steps:
# 1. Convert text to lowercase
# 2. Remove punctuation
# 3. Remove special characters
# 4. Remove extra spaces
# ============================================================

def preprocess_text(text):
    # Convert to lowercase
    text = text.lower()

    # Keep only English letters and spaces
    text = re.sub(r'[^a-z\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text


# Apply preprocessing to all 5 documents
preprocessed_documents = {}

for filename, text in documents.items():
    preprocessed_documents[filename] = preprocess_text(text)


# Display the results
print("=" * 70)
print("PREPROCESSING RESULTS")
print("=" * 70)

for filename in documents:

    print(f"\nDOCUMENT: {filename}")

    print("\nBefore preprocessing:")
    print(documents[filename][:300])

    print("\nAfter preprocessing:")
    print(preprocessed_documents[filename][:300])

    print("-" * 70)

print("\nPreprocessing completed successfully!")
```

```python
# ============================================================
# CELL 5: TOKENIZATION
# ============================================================
# Tokenization means splitting the text into individual words.
# ============================================================

tokenized_documents = {}

for filename, text in preprocessed_documents.items():
    tokens = text.split()
    tokenized_documents[filename] = tokens


# Display tokenization results
print("=" * 70)
print("TOKENIZATION RESULTS")
print("=" * 70)

for filename, tokens in tokenized_documents.items():

    print(f"\nDOCUMENT: {filename}")
    print("Total tokens:", len(tokens))

    print("\nFirst 30 tokens:")
    print(tokens[:30])

    print("-" * 70)

print("\nTokenization completed successfully!")
```

```python
# ============================================================
# CELL 6: STOPWORD REMOVAL
# ============================================================
# Stopwords are common words that usually do not carry
# important meaning for Information Retrieval.
# ============================================================

stopword_removed_documents = {}

for filename, tokens in tokenized_documents.items():

    # Remove English stopwords
    filtered_tokens = [
        word for word in tokens
        if word not in ENGLISH_STOP_WORDS
    ]

    stopword_removed_documents[filename] = filtered_tokens


# Display results
print("=" * 70)
print("STOPWORD REMOVAL RESULTS")
print("=" * 70)

for filename in tokenized_documents:

    before = tokenized_documents[filename]
    after = stopword_removed_documents[filename]

    print(f"\nDOCUMENT: {filename}")

    print("Tokens before stopword removal:", len(before))
    print("Tokens after stopword removal :", len(after))

    print("\nFirst 30 tokens after stopword removal:")
    print(after[:30])

    print("-" * 70)

print("\nStopword removal completed successfully!")
```

```python
# ============================================================
# CELL 7: STEMMING
# ============================================================
# Stemming reduces words to their basic stem form.
# ============================================================

stemmer = PorterStemmer()

stemmed_documents = {}

for filename, tokens in stopword_removed_documents.items():

    stemmed_tokens = [
        stemmer.stem(word)
        for word in tokens
    ]

    stemmed_documents[filename] = stemmed_tokens


# Display stemming results
print("=" * 70)
print("STEMMING RESULTS")
print("=" * 70)

for filename in stopword_removed_documents:

    original_tokens = stopword_removed_documents[filename]
    stemmed_tokens = stemmed_documents[filename]

    print(f"\nDOCUMENT: {filename}")

    print("First 30 tokens before stemming:")
    print(original_tokens[:30])

    print("\nFirst 30 tokens after stemming:")
    print(stemmed_tokens[:30])

    print("-" * 70)


# ============================================================
# SHOW SOME STEMMING EXAMPLES
# ============================================================

print("\n" + "=" * 70)
print("STEMMING EXAMPLES")
print("=" * 70)

example_words = [
    "communication",
    "communicating",
    "communities",
    "connected",
    "marketing",
    "education",
    "educational",
    "security",
    "students",
    "learning"
]

for word in example_words:
    print(f"{word:20} → {stemmer.stem(word)}")


print("\nStemming completed successfully!")
```

```python
# ============================================================
# CELL 8: TERM FREQUENCY (TF)
# ============================================================
# TF = Number of occurrences of a term
#      --------------------------------
#      Total number of terms in document
# ============================================================

tf_documents = {}

for filename, tokens in stemmed_documents.items():

    # Count how many times each term occurs
    term_counts = Counter(tokens)

    # Total number of terms in the document
    total_terms = len(tokens)

    # Calculate TF for every term
    tf_values = {}

    for term, count in term_counts.items():
        tf_values[term] = count / total_terms

    tf_documents[filename] = tf_values


# ============================================================
# CREATE A COMPLETE TF TABLE
# ============================================================

# Get all unique terms from all 5 documents
all_terms = sorted(
    set(
        term
        for document in tf_documents.values()
        for term in document.keys()
    )
)

# Create DataFrame
tf_table = pd.DataFrame(index=all_terms)

for filename in tf_documents:
    tf_table[filename] = [
        tf_documents[filename].get(term, 0)
        for term in all_terms
    ]


# ============================================================
# DISPLAY TF RESULTS
# ============================================================

print("=" * 70)
print("TERM FREQUENCY (TF)")
print("=" * 70)

print("\nFormula:")
print("TF = Number of occurrences of term / Total number of terms")

print("\nTotal number of unique terms:", len(all_terms))

print("\nComplete TF Table:")
display(tf_table.round(6))


# ============================================================
# TOP 20 TERMS BY TF FOR EACH DOCUMENT
# ============================================================

print("\n" + "=" * 70)
print("TOP 20 TERMS BY TF")
print("=" * 70)

for filename, tf_values in tf_documents.items():

    top_terms = sorted(
        tf_values.items(),
        key=lambda x: x[1],
        reverse=True
    )[:20]

    print(f"\nDOCUMENT: {filename}")
    print("-" * 50)

    for rank, (term, tf_value) in enumerate(top_terms, start=1):
        print(f"{rank:2}. {term:20} TF = {tf_value:.6f}")


print("\nTF calculation completed successfully!")
```

```python
# ============================================================
# CELL 9: DOCUMENT FREQUENCY (DF)
# ============================================================
# DF = Number of documents containing a particular term
# ============================================================

df_values = {}

# Total number of documents
N = len(stemmed_documents)

# Calculate DF for every term
for term in all_terms:

    document_count = 0

    for tokens in stemmed_documents.values():

        # Check whether the term appears in the document
        if term in set(tokens):
            document_count += 1

    df_values[term] = document_count


# ============================================================
# CREATE DF TABLE
# ============================================================

df_table = pd.DataFrame(
    list(df_values.items()),
    columns=["Term", "DF"]
)

# Sort by DF from highest to lowest
df_table = df_table.sort_values(
    by="DF",
    ascending=False
).reset_index(drop=True)


# ============================================================
# DISPLAY DF RESULTS
# ============================================================

print("=" * 70)
print("DOCUMENT FREQUENCY (DF)")
print("=" * 70)

print("\nNumber of documents (N):", N)

print("\nFormula:")
print("DF(term) = Number of documents containing the term")

print("\nComplete DF Table:")
display(df_table)


# ============================================================
# SHOW SOME EXAMPLES
# ============================================================

print("\n" + "=" * 70)
print("DF EXAMPLES")
print("=" * 70)

for _, row in df_table.head(20).iterrows():
    print(f"{row['Term']:20} → DF = {row['DF']}")


print("\nDF calculation completed successfully!")
```

```python
# ============================================================
# CELL 10: INVERSE DOCUMENT FREQUENCY (IDF)
# ============================================================
# IDF = log(N / DF)
#
# N  = Total number of documents
# DF = Number of documents containing the term
# ============================================================

# N is already calculated in Cell 9
print("Total number of documents (N):", N)


# ============================================================
# CALCULATE IDF FOR EVERY TERM
# ============================================================

idf_values = {}

for term in all_terms:

    df = df_values[term]

    # IDF formula
    idf = math.log(N / df)

    idf_values[term] = idf


# ============================================================
# CREATE IDF TABLE
# ============================================================

idf_table = pd.DataFrame(
    [
        [term, df_values[term], idf_values[term]]
        for term in all_terms
    ],
    columns=["Term", "DF", "IDF"]
)

# Sort by IDF from highest to lowest
idf_table = idf_table.sort_values(
    by="IDF",
    ascending=False
).reset_index(drop=True)


# ============================================================
# DISPLAY IDF RESULTS
# ============================================================

print("\n" + "=" * 70)
print("INVERSE DOCUMENT FREQUENCY (IDF)")
print("=" * 70)

print("\nFormula:")
print("IDF = log(N / DF)")

print("\nWhere:")
print("N =", N)
print("DF = Number of documents containing the term")

print("\nComplete IDF Table:")
display(idf_table.round(6))


# ============================================================
# SHOW IDF EXAMPLES
# ============================================================

print("\n" + "=" * 70)
print("IDF EXAMPLES")
print("=" * 70)

for _, row in idf_table.head(20).iterrows():

    term = row["Term"]
    df = int(row["DF"])
    idf = row["IDF"]

    print(
        f"{term:20} → "
        f"DF = {df} → "
        f"IDF = log({N}/{df}) = {idf:.6f}"
    )


# ============================================================
# IMPORTANT OBSERVATION
# ============================================================

print("\n" + "=" * 70)
print("IMPORTANT IDF OBSERVATION")
print("=" * 70)

print("""
If a term appears in all 5 documents:

IDF = log(5 / 5)
    = log(1)
    = 0

Therefore, terms appearing in many documents
have lower IDF.

Terms appearing in fewer documents
have higher IDF.
""")


print("\nIDF calculation completed successfully!")
```

```python
# ============================================================
# CELL 11: TF-IDF CALCULATION
# ============================================================
# TF-IDF = TF × IDF
#
# TF  → calculated in Cell 8
# IDF → calculated in Cell 10
# ============================================================

tfidf_table = pd.DataFrame(index=all_terms)

# Calculate TF-IDF for every term in every document
for filename in stemmed_documents:

    tfidf_values = []

    for term in all_terms:

        # Get TF value
        tf = tf_documents[filename].get(term, 0)

        # Get IDF value
        idf = idf_values[term]

        # TF-IDF formula
        tfidf = tf * idf

        tfidf_values.append(tfidf)

    tfidf_table[filename] = tfidf_values


# ============================================================
# DISPLAY COMPLETE TF-IDF TABLE
# ============================================================

print("=" * 70)
print("TF-IDF RESULTS")
print("=" * 70)

print("\nFormula:")
print("TF-IDF = TF × IDF")

print("\nComplete TF-IDF Table:")
display(tfidf_table.round(6))


# ============================================================
# TOP 10 TF-IDF TERMS FOR EACH DOCUMENT
# ============================================================

print("\n" + "=" * 70)
print("TOP 10 TF-IDF TERMS FOR EACH DOCUMENT")
print("=" * 70)

for filename in stemmed_documents:

    top_terms = (
        tfidf_table[filename]
        .sort_values(ascending=False)
        .head(10)
    )

    print(f"\nDOCUMENT: {filename}")
    print("-" * 50)

    for rank, (term, score) in enumerate(top_terms.items(), start=1):

        print(
            f"{rank:2}. {term:20} "
            f"TF-IDF = {score:.6f}"
        )


print("\nTF-IDF calculation completed successfully!")
```

```python
# ============================================================
# CELL 12: USER QUERY PROCESSING
# ============================================================

print("=" * 70)
print("INFORMATION RETRIEVAL - QUERY")
print("=" * 70)

# Ask the user to enter a search query
query = input("\nEnter your search query: ")

print("\nOriginal Query:")
print(query)


# ============================================================
# PREPROCESS THE QUERY
# ============================================================

# Lowercase
query_processed = query.lower()

# Remove punctuation and special characters
query_processed = re.sub(r'[^a-z\s]', ' ', query_processed)

# Remove extra spaces
query_processed = re.sub(r'\s+', ' ', query_processed).strip()


# ============================================================
# TOKENIZATION
# ============================================================

query_tokens = query_processed.split()


# ============================================================
# STOPWORD REMOVAL
# ============================================================

query_without_stopwords = [
    word for word in query_tokens
    if word not in ENGLISH_STOP_WORDS
]


# ============================================================
# STEMMING
# ============================================================

query_stemmed = [
    stemmer.stem(word)
    for word in query_without_stopwords
]


# ============================================================
# DISPLAY QUERY PROCESSING
# ============================================================

print("\n" + "=" * 70)
print("QUERY PROCESSING")
print("=" * 70)

print("\nOriginal Query:")
print(query)

print("\nAfter Preprocessing:")
print(query_processed)

print("\nAfter Tokenization:")
print(query_tokens)

print("\nAfter Stopword Removal:")
print(query_without_stopwords)

print("\nAfter Stemming:")
print(query_stemmed)

print("\nQuery processing completed successfully!")
```

```python
# ============================================================
# CELL 13: QUERY TF AND QUERY TF-IDF
# ============================================================

print("=" * 70)
print("QUERY TF AND QUERY TF-IDF")
print("=" * 70)

# ------------------------------------------------------------
# 1. Calculate Query TF
# ------------------------------------------------------------

query_term_counts = Counter(query_stemmed)

# Total number of terms in the processed query
query_total_terms = len(query_stemmed)

query_tf = {}

if query_total_terms > 0:

    for term, count in query_term_counts.items():
        query_tf[term] = count / query_total_terms

else:
    print("The query contains no valid terms.")


# ------------------------------------------------------------
# 2. Calculate Query TF-IDF
# ------------------------------------------------------------

query_tfidf = {}

for term, tf in query_tf.items():

    # Use IDF calculated from the 5 documents
    if term in idf_values:
        idf = idf_values[term]
    else:
        # If the query term does not occur in any document,
        # its document frequency is 0 and we assign IDF = 0
        idf = 0

    # TF-IDF = TF × IDF
    query_tfidf[term] = tf * idf


# ------------------------------------------------------------
# 3. Display Query TF
# ------------------------------------------------------------

print("\nQuery:")
print(query)

print("\nProcessed Query:")
print(query_stemmed)

print("\nQuery TF:")
print("-" * 50)

for term, value in query_tf.items():
    print(f"{term:20} TF = {value:.6f}")


# ------------------------------------------------------------
# 4. Display Query TF-IDF
# ------------------------------------------------------------

print("\nQuery TF-IDF:")
print("-" * 50)

for term, value in query_tfidf.items():
    print(f"{term:20} TF-IDF = {value:.6f}")


print("\nQuery TF and Query TF-IDF calculated successfully!")
```

```python
# ============================================================
# CELL 14: COSINE SIMILARITY AND DOCUMENT RANKING
# ============================================================

print("=" * 70)
print("COSINE SIMILARITY - DOCUMENT RETRIEVAL")
print("=" * 70)


# ------------------------------------------------------------
# 1. Create the Query TF-IDF Vector
# ------------------------------------------------------------

query_vector = np.array([
    query_tfidf.get(term, 0)
    for term in all_terms
])


# ------------------------------------------------------------
# 2. Calculate Cosine Similarity
#    between Query and each Document
# ------------------------------------------------------------

similarity_results = {}

for filename in tfidf_table.columns:

    # Get the document TF-IDF vector
    document_vector = tfidf_table[filename].values

    # Calculate dot product
    dot_product = np.dot(query_vector, document_vector)

    # Calculate magnitudes
    query_magnitude = np.linalg.norm(query_vector)
    document_magnitude = np.linalg.norm(document_vector)

    # Avoid division by zero
    if query_magnitude == 0 or document_magnitude == 0:
        similarity = 0
    else:
        similarity = (
            dot_product /
            (query_magnitude * document_magnitude)
        )

    similarity_results[filename] = similarity


# ------------------------------------------------------------
# 3. Rank Documents
# ------------------------------------------------------------

ranked_results = sorted(
    similarity_results.items(),
    key=lambda x: x[1],
    reverse=True
)


# ------------------------------------------------------------
# 4. Display Results
# ------------------------------------------------------------

print("\nQuery:")
print(query)

print("\nProcessed Query:")
print(query_stemmed)

print("\n" + "=" * 70)
print("SEARCH RESULTS")
print("=" * 70)

for rank, (filename, score) in enumerate(ranked_results, start=1):

    print(
        f"\nRank {rank}: {filename}"
    )

    print(
        f"Cosine Similarity: {score:.6f}"
    )


# ------------------------------------------------------------
# 5. Display Most Relevant Document
# ------------------------------------------------------------

most_relevant_document = ranked_results[0][0]
highest_score = ranked_results[0][1]

print("\n" + "=" * 70)
print("MOST RELEVANT DOCUMENT")
print("=" * 70)

print("\nQuery:", query)

print("Document:", most_relevant_document)

print(f"Similarity Score: {highest_score:.6f}")


# ------------------------------------------------------------
# 6. Create a Ranking Table
# ------------------------------------------------------------

ranking_table = pd.DataFrame(
    ranked_results,
    columns=["Document", "Cosine Similarity"]
)

ranking_table.insert(
    0,
    "Rank",
    range(1, len(ranking_table) + 1)
)

print("\n" + "=" * 70)
print("DOCUMENT RANKING TABLE")
print("=" * 70)

display(ranking_table.round(6))


print("\nDocument retrieval completed successfully!")
```

```python
# ============================================================
# CELL 15: FINAL IR SUMMARY AND CONCLUSION
# ============================================================

print("\n")
print("=" * 80)
print("              INFORMATION RETRIEVAL SYSTEM")
print("              FINAL SUMMARY")
print("=" * 80)

print("\n1. DOCUMENT COLLECTION")
print("-" * 80)
print(f"Number of documents processed: {len(documents)}")

for i, filename in enumerate(documents.keys(), start=1):
    print(f"{i}. {filename}")


print("\n2. PREPROCESSING")
print("-" * 80)
print("✓ Lowercase conversion")
print("✓ Punctuation and special-character removal")
print("✓ Extra-space removal")


print("\n3. TOKENIZATION")
print("-" * 80)
print("✓ Documents divided into individual words")


print("\n4. STOPWORD REMOVAL")
print("-" * 80)
print("✓ Common English stopwords removed")


print("\n5. STEMMING")
print("-" * 80)
print("✓ Words converted to their stem forms using Porter Stemmer")


print("\n6. TERM FREQUENCY (TF)")
print("-" * 80)
print("Formula:")
print("TF = Number of occurrences of term / Total number of terms")
print("✓ TF calculated manually for all documents")


print("\n7. DOCUMENT FREQUENCY (DF)")
print("-" * 80)
print("Formula:")
print("DF = Number of documents containing the term")
print("✓ DF calculated for all terms")


print("\n8. INVERSE DOCUMENT FREQUENCY (IDF)")
print("-" * 80)
print("Formula:")
print("IDF = log(N / DF)")
print(f"Where N = {N}")
print("✓ IDF calculated manually")


print("\n9. TF-IDF")
print("-" * 80)
print("Formula:")
print("TF-IDF = TF × IDF")
print("✓ TF-IDF calculated manually for all 5 documents")


print("\n10. USER QUERY")
print("-" * 80)
print("Original Query:")
print(query)

print("\nProcessed Query:")
print(query_stemmed)


print("\n11. DOCUMENT RETRIEVAL")
print("-" * 80)
print("Cosine Similarity was used to compare the query")
print("with the TF-IDF vector of each document.")


print("\n" + "=" * 80)
print("                 FINAL SEARCH RANKING")
print("=" * 80)

for rank, (filename, score) in enumerate(ranked_results, start=1):
    print(
        f"{rank}. {filename:<45} "
        f"Score = {score:.6f}"
    )


print("\n" + "=" * 80)
print("                 MOST RELEVANT DOCUMENT")
print("=" * 80)

print(f"\nQuery: {query}")
print(f"Document: {most_relevant_document}")
print(f"Cosine Similarity: {highest_score:.6f}")


print("\n" + "=" * 80)
print("                      CONCLUSION")
print("=" * 80)

print("""
This Information Retrieval system processes five social-media
documents through preprocessing, tokenization, stopword removal,
and stemming.

Term Frequency (TF) was calculated to measure how frequently a
term occurs in each document.

Document Frequency (DF) was calculated to determine how many
documents contain each term.

Inverse Document Frequency (IDF) was calculated using:

IDF = log(N / DF)

TF-IDF was then calculated using:

TF-IDF = TF × IDF

The user can enter a search query, which is processed using the
same preprocessing steps. Query TF-IDF is calculated and cosine
similarity is used to compare the query with each document.

Finally, the documents are ranked according to their similarity
scores, and the document with the highest score is identified as
the most relevant document for the query.
""")

print("=" * 80)
print("                     COMPLETED SUCCESSFULLY")
print("=" * 80)
```
